# SVG Animation Demo

This notebook demonstrates `data-animate` style SVG animation in a mystmd jupyterbook.

The animation below runs via inline JavaScript — the same SVG.js library used by the Reveal.js slideshow extension.

```python
from IPython.display import display, HTML
display(HTML("""
<script src='https://cdn.jsdelivr.net/npm/@svgdotjs/svg.js@3.1.2/dist/svg.min.js'></script>
<svg id='anim1' xmlns='http://www.w3.org/2000/svg' width='300' height='300'>
  <circle id='c1' cx='150' cy='150' r='60' style='fill:steelblue'/>
</svg>
<script>
(function() {
  function run() {
    if (typeof SVG === 'undefined') { setTimeout(run, 200); return; }
    var draw = SVG('#anim1');
    var c = draw.findOne('#c1');
    var c_el = c.node;
    c.animate(1500).attr({ r: 100 }).loop(true, true);
    var toggle = true;
    setInterval(function() {
      c_el.style.fill = toggle ? 'tomato' : 'steelblue';
      toggle = !toggle;
    }, 1500);
  }
  run();
})();
"""))
```
